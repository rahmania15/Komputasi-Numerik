PENYELESAIAN SISTEM PERSAMAAN LINEAR
TIGA VARIABEL (SPLTV)

Metode: Eliminasi Gauss-Jordan
Bahasa Pemrograman: Python

```python
# Membuat matriks augmented SPLTV
# Contoh persamaan:
# 2x + y + z = 7
# x + 2y + z = 8
# x + y + 2z = 9

A = [
    [2, 1, 1, 7],
    [1, 2, 1, 8],
    [1, 1, 2, 9]
]

# Menampilkan matriks awal
print("Matriks Augmented:")
for baris in A:
    print(baris)

    # Menentukan jumlah baris matriks
n = len(A)

# Melakukan eliminasi pada setiap kolom
for i in range(n):

    # Memilih pivot terbesar untuk menghindari
    # pembagian dengan nilai yang terlalu kecil
    pivot = max(range(i, n),
                key=lambda r: abs(A[r][i]))

    # Menukar posisi baris jika diperlukan
    A[i], A[pivot] = A[pivot], A[i]

    # Memeriksa apakah pivot bernilai nol
    if abs(A[i][i]) < 1e-12:
        raise ValueError("Sistem tidak memiliki solusi unik.")

    # Membuat nilai pivot menjadi 1
    pembagi = A[i][i]
    A[i] = [nilai / pembagi for nilai in A[i]]

    # Mengubah nilai kolom lainnya menjadi nol
    for j in range(n):
        if j != i:
            faktor = A[j][i]
            A[j] = [
                A[j][k] - faktor * A[i][k]
                for k in range(n + 1)
            ]

print("Matriks setelah eliminasi:")
for baris in A:
    print([round(nilai, 4) for nilai in baris])

    # Mengambil hasil akhir dari matriks
x = A[0][3]
y = A[1][3]
z = A[2][3]

# Menampilkan hasil
print("HASIL PENYELESAIAN SPLTV")
print("Nilai x =", round(x, 4))
print("Nilai y =", round(y, 4))
print("Nilai z =", round(z, 4))

# Memeriksa apakah solusi memenuhi persamaan awal
print("PENGUJIAN HASIL")

print("Persamaan 1:", 2*x + y + z)
print("Persamaan 2:", x + 2*y + z)
print("Persamaan 3:", x + y + 2*z)
```
